# Genome-wide selection scan with PBS

**Purpose.** The companion exercise builds PBS from raw reads on a single region. This one
takes PBS **already computed genome-wide** and asks the question you actually want answered:
across the whole genome, which regions stand out, and what genes are in them? A scan is only
useful if you can go from a peak to a gene to a plausible story — and if you can say why most
peaks are not stories at all.

**What you will do**
 - choose three populations, deciding which one's branch you are testing
 - plot PBS across the whole genome as a Manhattan plot
 - zoom into the most extreme regions chromosome by chromosome
 - identify the gene under a peak and look up what it does
 - check a known case, the lactase region, and see whether the scan recovers it

**The data.** Four populations from the **1000 Genomes Project**, with PBS precomputed in
50 kb windows across the genome:

| Code | Population |
|---|---|
| NAT | Native American (Peru and Mexico) |
| CHB | Han Chinese in Beijing |
| CEU | Utah residents with northern/western European ancestry |
| YRI | Yoruba in Ibadan, Nigeria |

**Called genotypes.** The PBS values are supplied precomputed because a genome-wide scan
takes far longer than a practical session — the computation itself is what the companion
exercise does from scratch.

**The order of the three populations matters.** PBS gives a branch length for the *first*
population you name, with the other two as comparisons. Choosing CEU, CHB, YRI tests the
European branch; reordering them tests a different question.

**Before this** do [SFS, Fst and PBS](sfs_fst_pbs_human.ipynb), which computes PBS from
sequencing data on one region so that these genome-wide values are not a black box.
